```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/supply_chain.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("Duplicate records:", df.duplicated().sum())
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print(df["Order_Date"].dtype)
```

```python
df.describe()
```

```python
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sale:", df["Sales"].max())
print("Minimum Sale:", df["Sales"].min())
```

```python
low_stock = df[df["Stock"] < df["Reorder_Level"]]

print("Low Stock Records:", len(low_stock))

low_stock[
    [
        "Product_Name",
        "Stock",
        "Reorder_Level"
    ]
].head(20)
```

```python
delayed_orders = df[
    df["Delivery_Status"] == "Delayed"
]

print("Delayed Orders:", len(delayed_orders))
```

```python
delayed_orders[
    [
        "Order_ID",
        "Supplier_Name",
        "Delivery_Days",
        "Delivery_Status"
    ]
].head(20)
```

```python
df = df.drop_duplicates()

df.to_csv(
    "../data/cleaned_supply_chain.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
