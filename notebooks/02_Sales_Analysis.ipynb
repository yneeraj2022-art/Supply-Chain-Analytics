```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/cleaned_supply_chain.csv")

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print("Dataset loaded successfully!")
print("Rows:", len(df))
```

```python
total_sales = df["Sales"].sum()

print("Total Sales:", total_sales)
```

```python
total_orders = df["Order_ID"].nunique()

print("Total Orders:", total_orders)
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
product_sales = (
    df.groupby("Product_Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

product_sales
```

```python
top_products = product_sales.head(5)

print("Top 5 Products:")
print(top_products)
```

```python
plt.figure(figsize=(10, 5))

product_sales.plot(kind="bar")

plt.title("Product-wise Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
plt.figure(figsize=(7, 7))

category_sales.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Category-wise Sales Distribution")
plt.ylabel("")

plt.show()
```

```python
df["Month"] = df["Order_Date"].dt.month
```

```python
monthly_sales = (
    df.groupby("Month")["Sales"]
    .sum()
)

monthly_sales
```

```python
plt.figure(figsize=(10, 5))

monthly_sales.plot(
    kind="line",
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(range(1, 13))

plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
supplier_sales = (
    df.groupby("Supplier_Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

supplier_sales
```

```python
plt.figure(figsize=(10, 5))

supplier_sales.plot(kind="bar")

plt.title("Supplier-wise Sales")
plt.xlabel("Supplier")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
order_status = df["Order_Status"].value_counts()

print(order_status)
```

```python
delivery_status = df["Delivery_Status"].value_counts()

print(delivery_status)
```

```python

plt.figure(figsize=(7, 5))

delivery_status.plot(kind="bar")

plt.title("Delivery Status")
plt.xlabel("Status")
plt.ylabel("Number of Orders")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
average_delivery = df["Delivery_Days"].mean()

print(
    "Average Delivery Time:",
    round(average_delivery, 2),
    "days"
)
```

```python
print("========== SALES SUMMARY ==========")

print("Total Sales:", df["Sales"].sum())

print("Total Orders:", df["Order_ID"].nunique())

print("Total Quantity Sold:", df["Quantity"].sum())

print(
    "Average Order Value:",
    round(df["Sales"].mean(), 2)
)

print(
    "Average Delivery Days:",
    round(df["Delivery_Days"].mean(), 2)
)

print("===================================")
```
