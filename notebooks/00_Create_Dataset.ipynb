```python
import pandas as pd
import random
from datetime import datetime, timedelta

print("Libraries imported successfully!")
```

```python
products = [
    ("P001", "Laptop", "Electronics", 50000),
    ("P002", "Mobile Phone", "Electronics", 25000),
    ("P003", "Tablet", "Electronics", 18000),
    ("P004", "Monitor", "Electronics", 12000),
    ("P005", "Keyboard", "Accessories", 1500),
    ("P006", "Mouse", "Accessories", 800),
    ("P007", "Headphones", "Accessories", 2500),
    ("P008", "Printer", "Office", 15000),
    ("P009", "Office Chair", "Furniture", 7000),
    ("P010", "Desk", "Furniture", 10000)
]

suppliers = [
    ("S001", "Tech Supplier Ltd"),
    ("S002", "Global Electronics"),
    ("S003", "Smart Supply Co"),
    ("S004", "Prime Distributors"),
    ("S005", "Reliable Suppliers")
]

print("Products:", len(products))
print("Suppliers:", len(suppliers))
```

```python
delivery_statuses = [
    "On Time",
    "Delayed",
    "On Time",
    "On Time",
    "Delayed"
]

order_statuses = [
    "Delivered",
    "Delivered",
    "Delivered",
    "Pending",
    "Cancelled"
]

data = []

start_date = datetime(2025, 1, 1)

for i in range(1, 1001):

    product_id, product_name, category, unit_price = random.choice(products)

    supplier_id, supplier_name = random.choice(suppliers)

    quantity = random.randint(1, 20)

    sales = quantity * unit_price

    stock = random.randint(5, 100)

    reorder_level = random.randint(10, 30)

    delivery_days = random.randint(2, 12)

    delivery_status = random.choice(delivery_statuses)

    order_status = random.choice(order_statuses)

    order_date = start_date + timedelta(
        days=random.randint(0, 364)
    )

    data.append([
        1000 + i,
        order_date.strftime("%Y-%m-%d"),
        product_id,
        product_name,
        category,
        supplier_id,
        supplier_name,
        quantity,
        unit_price,
        sales,
        stock,
        reorder_level,
        delivery_days,
        delivery_status,
        order_status
    ])

print("Records generated:", len(data))
```

```python
columns = [
    "Order_ID",
    "Order_Date",
    "Product_ID",
    "Product_Name",
    "Category",
    "Supplier_ID",
    "Supplier_Name",
    "Quantity",
    "Unit_Price",
    "Sales",
    "Stock",
    "Reorder_Level",
    "Delivery_Days",
    "Delivery_Status",
    "Order_Status"
]

df = pd.DataFrame(data, columns=columns)

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.info()
```

```python
from pathlib import Path

# Current JupyterLab location
current_folder = Path.cwd()

print("Current folder:", current_folder)

# Find the data folder
if (current_folder / "data").exists():
    data_folder = current_folder / "data"
elif (current_folder.parent / "data").exists():
    data_folder = current_folder.parent / "data"
else:
    data_folder = current_folder / "data"
    data_folder.mkdir(parents=True, exist_ok=True)

# Save dataset
file_path = data_folder / "supply_chain.csv"

df.to_csv(file_path, index=False)

print("Dataset saved successfully!")
print("File location:", file_path)
```
